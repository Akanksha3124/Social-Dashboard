```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (12, 5)

yt = pd.read_csv('../data/cleaned/youtube_clean.csv')

print('=== YouTube Overview ===')
print(f'Total videos    : {len(yt)}')
print(f'Total views     : {yt["views"].sum():,}')
print(f'Total likes     : {yt["likes"].sum():,}')
print(f'Total comments  : {yt["comments"].sum():,}')
print(f'Avg engagement  : {yt["engagement_rate"].mean():.2%}')
print()
```

```python
top10 = yt.nlargest(10, 'views')[['title', 'views', 'likes', 'comments', 'engagement_rate']]
top10['title'] = top10['title'].str[:40]
print(top10.to_string(index=False))

fig, ax = plt.subplots()
ax.barh(top10['title'], top10['views'], color='steelblue')
ax.set_xlabel('Views')
ax.set_title('Top 10 YouTube videos by views')
plt.tight_layout()
plt.savefig('../data/processed/top10_views.png', dpi=150)
plt.show()
```

```python
day_order = ['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday']
day_eng = yt.groupby('day_of_week')['engagement_rate'].mean().reindex(day_order)

fig, ax = plt.subplots()
bars = ax.bar(day_eng.index, day_eng.values, color='#4C72B0')
best_day = day_eng.idxmax()
bars[day_order.index(best_day)].set_color('#DD5533')
ax.set_title('Average engagement rate by day (YouTube)')
ax.set_ylabel('Engagement rate')
ax.set_xlabel('Day')
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig('../data/processed/day_engagement.png', dpi=150)
plt.show()
print(f'Best day to post: {best_day}')
```

```python
pivot = yt.groupby(['day_of_week', 'hour'])['engagement_rate'].mean().unstack(fill_value=0)
pivot = pivot.reindex(day_order)

fig, ax = plt.subplots(figsize=(16, 5))
sns.heatmap(pivot, cmap='YlOrRd', annot=True, fmt='.2f', linewidths=0.5, ax=ax)
ax.set_title('Engagement rate heatmap — Day vs Hour (YouTube)')
ax.set_xlabel('Hour of day (0–23)')
ax.set_ylabel('Day of week')
plt.tight_layout()
plt.savefig('../data/processed/heatmap.png', dpi=150)
plt.show()

best_hour = yt.groupby('hour')['engagement_rate'].mean().idxmax()
print(f'Best hour to post: {best_hour}:00')
```

```python
fig, ax = plt.subplots()
ax.scatter(yt['views'], yt['likes'], alpha=0.6, color='steelblue', edgecolors='white', s=60)
ax.set_xlabel('Views')
ax.set_ylabel('Likes')
ax.set_title('Views vs Likes (YouTube)')
plt.tight_layout()
plt.savefig('../data/processed/views_vs_likes.png', dpi=150)
plt.show()
```

```python
yt_best_day  = yt.groupby('day_of_week')['engagement_rate'].mean().idxmax()
yt_best_hour = yt.groupby('hour')['engagement_rate'].mean().idxmax()

print('====== CONTENT STRATEGY SUMMARY ======')
print(f'YouTube — Post on : {yt_best_day} at {yt_best_hour}:00')
print()
print('Top YouTube video:')
top = yt.nlargest(1, 'engagement_rate').iloc[0]
print(f'  Title      : {top["title"][:60]}')
print(f'  Views      : {top["views"]:,}')
print(f'  Engagement : {top["engagement_rate"]:.2%}')
```
